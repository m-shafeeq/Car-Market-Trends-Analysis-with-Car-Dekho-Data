```python
data['Depreciation'] = data['Present_Price'] - data['Selling_Price']

most_dep = data.loc[data['Depreciation'].idxmax()]
least_dep = data.loc[data['Depreciation'].idxmin()]

print("Most depreciated:\n", most_dep[['Car_Name', 'Present_Price', 'Selling_Price', 'Depreciation']])
print("\nLeast depreciated:\n", least_dep[['Car_Name', 'Present_Price', 'Selling_Price', 'Depreciation']])
```

```python
# The dataset has no separate "brand" column, so pull the first word of the name.
data['Brand'] = data['Car_Name'].apply(lambda x: x.split()[0])
data['Depreciation_pct'] = (data['Present_Price'] - data['Selling_Price']) / data['Present_Price'] * 100

brand_dep = data.groupby('Brand')['Depreciation_pct'].mean().sort_values()

print("Least depreciation (top 5):")
print(brand_dep.head(5))

print("\nMost depreciation (top 5):")
print(brand_dep.tail(5))
```

```python
# Correlate each numeric factor against Depreciation_pct
corr = data[['Year', 'Kms_Driven', 'Owner', 'Depreciation_pct']].corr()
print(corr['Depreciation_pct'])
```

```python
data['Age'] = 2018 - data['Year']    # 2018 = latest year in this dataset

corr2 = data[['Age', 'Kms_Driven', 'Selling_Price']].corr()
print(corr2['Selling_Price'])
```

```python
after_2014 = data[data['Year'] > 2014]
print(len(after_2014))
after_2014.head()
```

```python
bike_brands = ['Activa', 'Bajaj', 'Hero', 'Honda', 'Hyosung', 'KTM',
               'Mahindra Mojo', 'Royal Enfield', 'Suzuki Access', 'TVS',
               'UM Renegade', 'Yamaha']

def is_bike(name):
    return any(name.strip().startswith(b) for b in bike_brands)

data['Vehicle_Type'] = data['Car_Name'].apply(lambda n: 'Two-Wheeler' if is_bike(n) else 'Car')

bikes = data[data['Vehicle_Type'] == 'Two-Wheeler']
cars = data[data['Vehicle_Type'] == 'Car']

print("Two-wheelers:", len(bikes))
print("Cars:", len(cars))
```

```python
bikes.head(10)
```

```python
oldest_bike = bikes.loc[bikes['Year'].idxmin()]
print(oldest_bike[['Car_Name', 'Year']])
```

```python
newest_bike = bikes.loc[bikes['Year'].idxmax()]
print(newest_bike[['Car_Name', 'Year']])
```

```python
most_sold_bike = bikes['Car_Name'].value_counts()
print(most_sold_bike.head(1))
```

```python
# Fit a simple "expected price" model from Age + Kms_Driven, then look
# for the bike that sold well ABOVE what the model predicts.
X = np.c_[np.ones(len(bikes)), bikes['Age'], bikes['Kms_Driven']]
y = bikes['Selling_Price'].values

coef, *_ = np.linalg.lstsq(X, y, rcond=None)   # least-squares regression
predicted = X @ coef
residual = y - predicted                        # actual - expected

bikes_check = bikes.copy()
bikes_check['Expected_Price'] = predicted
bikes_check['Above_Expected'] = residual

best_deal_bike = bikes_check.loc[bikes_check['Above_Expected'].idxmax()]
print(best_deal_bike[['Car_Name', 'Year', 'Selling_Price', 'Expected_Price', 'Above_Expected']])
```

```python
# Same approach as Q21, but on the cars subset
X = np.c_[np.ones(len(cars)), cars['Age'], cars['Kms_Driven']]
y = cars['Selling_Price'].values

coef, *_ = np.linalg.lstsq(X, y, rcond=None)
predicted = X @ coef
residual = y - predicted

cars_check = cars.copy()
cars_check['Expected_Price'] = predicted
cars_check['Above_Expected'] = residual

best_deal_car = cars_check.loc[cars_check['Above_Expected'].idxmax()]
print(best_deal_car[['Car_Name', 'Year', 'Selling_Price', 'Expected_Price', 'Above_Expected']])
```
